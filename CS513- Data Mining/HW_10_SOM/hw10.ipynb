### Import and Handling Data

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from minisom import MiniSom

temp= pd.read_csv("wisc_bc_ContinuousVar.csv")
temp= temp.dropna()
f= temp.drop(columns=['diagnosis'])
d= temp['diagnosis']


```

### Normalizing Features

```python
sclTemp= StandardScaler()
norm= sclTemp.fit_transform(f)
```

### Traning and Handling Clusters with Self-Organizing Map (SOM)

```python
som= MiniSom(x=1, y=2, input_len=norm.shape[1], sigma=1.0, learning_rate=0.5)
som.train(norm, 100)
clL= np.array([som.winner(x) for x in norm])
all= pd.DataFrame({'SOM_Cluster': clL[:, 1], 'Diagnosis': d})
tabRes= pd.crosstab(all['SOM_Cluster'], all['Diagnosis'])
print(tabRes)
```
