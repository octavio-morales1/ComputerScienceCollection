1-Load the “breast-cancer-wisconsin.data.csv” from canvas into R/Python and perform the EDA analysis by:

```python
# Loading Data and Modules
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data= pd.read_csv('breast-cancer-wisconsin.csv')
data.head()
```

I. Summarizing each column (e.g. min, max, mean )

```python
summ= data.describe()
print(summ)
```

II. Identifying missing values

```python
missing= data.isna().sum()
print(missing)
```

III. Replacing the missing values with the “mean” of the column.

```python
data.replace("?", pd.NA, inplace=True)
data= data.apply(pd.to_numeric, errors='coerce')
data.fillna(data.mean(), inplace=True)
```

IV. Displaying the frequency table of “Class” vs. F6

```python
fr_table= pd.crosstab(data['Class'], data['F6'])
print(fr_table)
```

V. Displaying the scatter plot of F1 to F6, one pair at a time

```python
sns.pairplot(data, vars=["F1", "F2", "F3", "F4", "F5", "F6"], hue="Class", diag_kind='kde')
plt.suptitle("Scatter Plot Matrix: F1 to F6", y=1.02)
plt.show()
```

VI. Show histogram box plot for columns F7 to F9

```python
fig, ax= plt.subplots(2, 3, figsize=(15, 10))

# F7
ax[0, 0].hist(data['F7'], bins=10, color='skyblue', edgecolor='black')
ax[0, 0].set_title('Histogram: F7')
ax[1, 0].boxplot(data['F7'], vert=False)
ax[1, 0].set_title('Box Plot for F7')

# F8
ax[0, 1].hist(data['F8'], bins=10, color='salmon', edgecolor='black')
ax[0, 1].set_title('Histogram: F8')
ax[1, 1].boxplot(data['F8'], vert=False)
ax[1, 1].set_title('Box Plot: F8')

# F9
ax[0, 2].hist(data['F9'], bins=10, color='lightgreen', edgecolor='black')
ax[0, 2].set_title('Histogram: F9')
ax[1, 2].boxplot(data['F9'], vert=False)
ax[1, 2].set_title('Box Plot: F9')

plt.tight_layout()
plt.show()
```
